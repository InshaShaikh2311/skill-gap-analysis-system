```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np

# LOAD GAP DATA
gap_df = pd.read_csv("../data/processed/gap_analysis.csv")

# TAKE TOP SKILLS
top_skills = gap_df.sort_values("Industry_Demand", ascending=False).head(10)

predictions = []

for index,row in top_skills.iterrows():
    
    skill = row['Skill']
    demand = row['Industry_Demand']
    
    # create trend based on demand
    y = np.array([
        demand*0.6,
        demand*0.75,
        demand*0.9,
        demand
    ])
    
    years = np.array([2022,2023,2024,2025]).reshape(-1,1)

    model = LinearRegression()
    model.fit(years,y)

    future = model.predict([[2026]])

    predictions.append([skill, future[0]])

# SAVE RESULT
future_df = pd.DataFrame(predictions, columns=["Skill","Predicted_2026_Demand"])

future_df.to_csv("../data/processed/future_predictions.csv", index=False)

print(future_df)

print("03_future_prediction completed successfully")
```

```python
import matplotlib.pyplot as plt

import matplotlib.pyplot as plt
import os
import pandas as pd

# load predicted data
future_df = pd.read_csv("../data/processed/future_predictions.csv")

# take top 8 skills
top8 = future_df.sort_values(
    "Predicted_2026_Demand",
    ascending=False
).head(8)

# create folder if not exists
os.makedirs("../static/charts", exist_ok=True)

plt.figure(figsize=(7,7))

plt.pie(
    top8['Predicted_2026_Demand'],
    labels=top8['Skill'],
    autopct='%1.1f%%'
)

plt.title("Future Skill Demand Distribution (2026)")

plt.tight_layout()

# SAVE image instead of show
plt.savefig("../static/charts/future_pie.png")

plt.close()

print("Future pie chart saved successfully")
```
