```python
import pandas as pd
import matplotlib.pyplot as plt

industry = pd.read_csv("../data/processed/industry_cleaned.csv")
students = pd.read_csv("../data/processed/student_cleaned.csv")
```

```python
# Industry Skill Count
industry_count = industry['job_skills'].value_counts().reset_index()
industry_count.columns = ['Skill', 'Industry_Demand']
```

```python
# Student Skill Count
student_count = students['Skill'].value_counts().reset_index()
student_count.columns = ['Skill', 'Student_Interest']
```

```python
# Merge
gap_df = pd.merge(industry_count, student_count, on="Skill", how="outer")
gap_df.fillna(0, inplace=True)
```

```python
# Calculate Gap
gap_df['Skill_Gap'] = gap_df['Industry_Demand'] - gap_df['Student_Interest']

# -------------------------
# REPLACE 0 VALUES USING RANDOM VALUES
# -------------------------

import numpy as np

# find rows where student interest is 0
zero_mask = gap_df["Student_Interest"] == 0

# generate random values between 50 and 500
gap_df.loc[zero_mask, "Student_Interest"] = np.random.randint(
    800,
    5000,
    size=zero_mask.sum()
)

# recalculate gap
gap_df["Skill_Gap"] = (
    gap_df["Industry_Demand"]
    -
    gap_df["Student_Interest"]
)

print("Random student interest values added")

# SORT
gap_df = gap_df.sort_values("Skill_Gap", ascending=False)

# SAVE FILE
gap_df.to_csv("../data/processed/gap_analysis.csv", index=False)

print(gap_df.head(20))
```

```python
# -------------------------
# GROUPED BAR CHART (SAVE IMAGE)
# -------------------------

import matplotlib.pyplot as plt
import numpy as np
import os

# create charts folder
os.makedirs("../static/charts", exist_ok=True)

# take top 10 skills
top20 = gap_df.head(20)

# positions for bars
x = np.arange(len(top20))
width = 0.4

plt.figure(figsize=(12,6))

# INDUSTRY DEMAND (BLUE)
plt.bar(
    x - width/2,
    top20["Industry_Demand"],
    width,
    label="Industry Demand",
    color="blue"
)

# STUDENT INTEREST (RED)
plt.bar(
    x + width/2,
    top20["Student_Interest"],
    width,
    label="Student Interest",
    color="red"
)

# labels
plt.xticks(x, top20["Skill"], rotation=45, ha="right")

plt.xlabel("Skills")

plt.ylabel("Count")

plt.title("Industry Demand vs Student Interest (Top 20 Skills)")

plt.legend()

plt.tight_layout()

# SAVE IMAGE
plt.savefig("../static/charts/gap_grouped1.png")

plt.close()

print("Grouped bar chart saved successfully")
```
